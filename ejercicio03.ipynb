```python
#1.
import requests

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()

nombres = []

for u in usuarios:
    if u["email"].endswith(".biz"):
        nombres.append(u["name"])

print("Usuarios con correo terminado en .biz:")

for n in nombres:
    print("-", n)
```

```python
#2.
import requests

nombre_pokemon = input("Ingrese el nombre de un Pokemon: ").lower()

url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    datos = respuesta.json()

    tipos = []
    for t in datos["types"]:
        tipos.append(t["type"]["name"])

    habilidades = []
    for h in datos["abilities"]:
        habilidades.append(h["ability"]["name"])

    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": tipos,
        "habilidades": habilidades
    }

    print("\nInformacion del Pokemon")
    print("Nombre:", pokemon["nombre"].capitalize())
    print("Altura:", pokemon["altura"])
    print("Peso:", pokemon["peso"])
    print("Tipos:", ", ".join(pokemon["tipos"]))

    if len(pokemon["tipos"]) > 1:
        print("Pokemon de tipo mixto")
    else:
        print("Pokemon de tipo unico")

    if pokemon["peso"] > 100:
        print("Pokemon de gran tamaño")
    else:
        print("Pokemon ligero")

    print("Habilidades:")
    for habilidad in pokemon["habilidades"]:
        print("-", habilidad)

    print("Experiencia base:", datos["base_experience"])
    print("Numero de movimientos:", len(datos["moves"]))

else:
    print("Pokemon no encontrado. Verifique el nombre ingresado.")
```
