```python
# 2.
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
print("Temperaturas:", temperaturas)

temperaturas.append(23.7)
print("Después de agregar al final:", temperaturas)

temperaturas.pop(0)
print("Después de eliminar la primera:", temperaturas)

temperaturas.insert(1, 22.0)
print("Después de insertar en la segunda posición:", temperaturas)

print("Ordenada:", sorted(temperaturas))

diferencia = max(temperaturas) - min(temperaturas)
print(f"Diferencia entre la más alta y la más baja: {diferencia:.2f}")

promedio = sum(temperaturas) / len(temperaturas)
print(f"Promedio: {promedio:.2f}")

print("Índice de 23.7:", temperaturas.index(23.7))

print("De la 2da a la 4ta posición:", temperaturas[1:4])

print("Tres primeras:", temperaturas[:3])
```

```python
#3.
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10}
]

print("Primer producto:", inventario[0]["nombre"])

nuevo = {"nombre": "Arroz", "precio": 2500, "stock": 20}
inventario.append(nuevo)

for p in inventario:
    if p["nombre"] == "Leche":
        p["stock"] = p["stock"] - 2

total = 0

for p in inventario:
    total = total + p["precio"] * p["stock"]

print("Valor total del inventario:", total)

print("Productos con stock menor a 10:")

for p in inventario:
    if p["stock"] < 10:
        print(p["nombre"])


for p in inventario:
    p["precio"] = p["precio"] * 1.10

for p in inventario:
    p["precio"] = round(p["precio"], 2)

busqueda = input("Ingrese el nombre del producto que desea buscar: ")
encontrado = False

for p in inventario:
    if p["nombre"].lower() == busqueda.lower():
        print("Producto encontrado:", p)
        encontrado = True

if encontrado == False:
    print("El producto no existe en el inventario")

for p in inventario[:]:
    if p["stock"] == 0:
        inventario.remove(p)

inventario.sort(key=lambda p: p["precio"])

print("\nInventario final:")

for p in inventario:
    print(p)
```

```python
#4.
rectangulo = (5, 3)

ancho = rectangulo[0]
alto = rectangulo[1]

area = ancho * alto
perimetro = 2 * (ancho + alto)

print("Rectangulo:", rectangulo)
print("Area:", area)
print("Perimetro:", perimetro)

try:
    rectangulo[0] = 10
except TypeError as error:
    print("Error:", error)
    print("No se puede cambiar un valor de una tupla porque es inmutable.")

posiciones = [(0, 0), (1, 2), (3, 4)]

distancia_total = 0

for i in range(len(posiciones) - 1):
    x1 = posiciones[i][0]
    y1 = posiciones[i][1]

    x2 = posiciones[i + 1][0]
    y2 = posiciones[i + 1][1]

    distancia = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5

    distancia_total = distancia_total + distancia

print("Distancia total recorrida:", round(distancia_total, 2))
```

```python
#5.
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True}
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

prestados = 0
disponibles = 0

for libro in biblioteca["libros"]:
    if libro["prestado"] == True:
        prestados = prestados + 1
    else:
        disponibles = disponibles + 1

print("Libros prestados:", prestados)
print("Libros disponibles:", disponibles)

nuevo_libro = {
    "titulo": "Cien años de soledad",
    "autor": "Gabriel García Márquez",
    "prestado": False
}

biblioteca["libros"].append(nuevo_libro)

for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True

print(biblioteca)
```
