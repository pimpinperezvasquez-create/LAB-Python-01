```python
#1.
import random

ganadas = 0
perdidas = 0

while True:
    numero = random.randint(1, 100)
    intentos = []
    encontrado = False

    for i in range(7):
        intento = int(input("Adivina el número entre 1 y 100: "))
        intentos.append(intento)

        if intento == numero:
            print("Correcto, ganaste")
            encontrado = True
            ganadas = ganadas + 1
            break
        elif intento < numero:
            print("Mayor")
        else:
            print("Menor")

    if encontrado == False:
        print("Perdiste. El número era:", numero)
        perdidas = perdidas + 1

    print("Historial de intentos:", intentos)
    print("Partidas ganadas:", ganadas)
    print("Partidas perdidas:", perdidas)

    otra = input("¿Quieres jugar otra vez? (s/n): ")

    if otra.lower() != "s":
        break
```

Siete intentos son fucientes debido a que la búsqueda binaria divide el rango en dos en cada intento. Con 7 intentos se pueden distinguir hasta (2^7 = 128) posibilidades, suficientes para encontrar un número entre 1 y 100 si se elige siempre el punto medio

```python
#2.
import random

tablero = []

for i in range(10):
    fila = []
    for j in range(10):
        fila.append(random.randint(0, 1))
    tablero.append(fila)

def contar_vecinos(tablero, fila, columna):
    vivos = 0

    for i in range(fila - 1, fila + 2):
        for j in range(columna - 1, columna + 2):
            if i >= 0 and i < 10 and j >= 0 and j < 10:
                if i != fila or j != columna:
                    vivos = vivos + tablero[i][j]

    return vivos

def mostrar_tablero(tablero):
    for fila in tablero:
        for celda in fila:
            if celda == 1:
                print("■", end=" ")
            else:
                print(".", end=" ")
        print()
    print()

for generacion in range(5):
    print("Generación", generacion + 1)
    mostrar_tablero(tablero)

    nuevo_tablero = []

    for i in range(10):
        fila = []

        for j in range(10):
            vecinos = contar_vecinos(tablero, i, j)

            if tablero[i][j] == 1:
                if vecinos == 2 or vecinos == 3:
                    fila.append(1)
                else:
                    fila.append(0)
            else:
                if vecinos == 3:
                    fila.append(1)
                else:
                    fila.append(0)

        nuevo_tablero.append(fila)

    tablero = nuevo_tablero
```

Usamos un tablero nuevo debido a que todas las celdas deben calcularse usando el estado de la generación anterior. Si modificamos el tablero original mientras lo recorremos, los cambios afectarían los cálculos de las celdas siguientes y el resultado sería incorrecto
