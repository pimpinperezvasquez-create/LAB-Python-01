```python
#1.
productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")

    if nombre == "fin":
        break

    while True:
        try:
            precio = float(input("Precio: "))
            cantidad = int(input("Cantidad: "))

            if precio < 0 or cantidad < 0:
                print("El precio y la cantidad no pueden ser negativos")
                continue

        except ValueError:
            print("Ingrese valores numéricos válidos")
            continue

        else:
            print("Datos ingresados correctamente")

        finally:
            print("Validación finalizada")

        break

    productos.append({
        "nombre": nombre,
        "precio": precio,
        "cantidad": cantidad
    })

total = 0

for p in productos:
    total = total + p["precio"] * p["cantidad"]

print("Inventario:", productos)
print("Total del inventario:", total)
```

```python
#2. a.
productos = []

try:
    with open("inventario.csv", "r", encoding="utf-8") as archivo:
        for linea in archivo:
            datos = linea.strip().split(",")
            producto = {
                "nombre": datos[0],
                "precio": float(datos[1]),
                "cantidad": int(datos[2])
            }
            productos.append(producto)

except FileNotFoundError:
    productos = []

print("Inventario cargado:", productos)

if len(productos) == 0:
    productos.append({"nombre": "Pan", "precio": 1500, "cantidad": 10})
    productos.append({"nombre": "Leche", "precio": 3500, "cantidad": 5})

with open("inventario.csv", "w", encoding="utf-8") as archivo:
    for p in productos:
        archivo.write(p["nombre"] + "," + str(p["precio"]) + "," + str(p["cantidad"]) + "\n")

print("Inventario guardado correctamente")
```

```python
#2. b.
import json

productos = []

try:
    with open("inventario.json", "r", encoding="utf-8") as archivo:
        productos = json.load(archivo)

except FileNotFoundError:
    productos = []

print("Inventario cargado:", productos)

if len(productos) == 0:
    productos.append({"nombre": "Pan", "precio": 1500, "cantidad": 10})
    productos.append({"nombre": "Leche", "precio": 3500, "cantidad": 5})

with open("inventario.json", "w", encoding="utf-8") as archivo:
    json.dump(productos, archivo, indent=4, ensure_ascii=False)

print("Inventario guardado correctamente")
```

**Ventaja de JSON:** permite guardar directamente listas y diccionarios, conserva mejor la estructura de los datos y facilita trabajar con información más compleja que la de un archivo CSV
