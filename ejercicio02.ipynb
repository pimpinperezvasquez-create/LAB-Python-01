```python
#1.
edad = int(input("Ingrese su edad: "))
tiene_licencia = input("¿Tiene licencia? (s/n): ")
sancionado = input("¿Está sancionado? (s/n): ")

if edad >= 18 and tiene_licencia == "s" and sancionado == "n":
    print("Puede conducir")
else:
    print("No puede conducir")
```

```python
#2.
n = int(input("Ingrese la cantidad de números: "))

suma = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        print("Número par:", i)

    suma = suma + i

print("La suma de los primeros", n, "números naturales es:", suma)


temperaturas = [20, 25, 18, 30, 22]

suma = 0

for temperatura in temperaturas:
    suma = suma + temperatura

promedio = suma / len(temperaturas)

print("Temperatura promedio:", promedio)
print("Temperaturas mayores al promedio:")

for temperatura in temperaturas:
    if temperatura > promedio:
        print(temperatura)
```

```python
#3.
contador = 5

while contador > 0:
    print("Contador:", contador)
    contador = contador - 1

numero = int(input("Ingrese un número (0 para terminar): "))

while numero != 0:
    print("Número ingresado:", numero)
    numero = int(input("Ingrese otro número (0 para terminar): "))

print("El ciclo ha terminado")
```

```python
#4.
suma_pares = 0
suma_impares = 0

for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

resultado = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}

print(resultado)
```

```python
#5.
productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")

    if nombre == "fin":
        break

    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))

    if precio < 0 or cantidad < 1:
        continue

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

total = 0

for p in productos:
    total = total + p["precio"] * p["cantidad"]

print("Total del inventario:", total)

if len(productos) > 0:
    producto_caro = max(productos, key=lambda p: p["precio"])
    print("Producto más caro:", producto_caro)

print("Inventario final:", productos)
```
