```python
#2.
def agregar_nota(notas, nueva):
    notas.append(nueva)

def reiniciar(notas):
    notas.clear()

mis_notas = [3.5, 4.0]

print("ID inicial:", id(mis_notas))

agregar_nota(mis_notas, 4.8)
print("Después de agregar:", mis_notas)
print("ID después de agregar:", id(mis_notas))

reiniciar(mis_notas)
print("Después de reiniciar:", mis_notas)
print("ID después de reiniciar:", id(mis_notas))
```

```python
#3.
cuenta = {"titular": "Ana", "saldo": 200000, "movimientos": []}

def consultar(cuenta):
    print("Saldo actual:", cuenta["saldo"])

def depositar(cuenta, monto):
    if monto > 0:
        cuenta["saldo"] = cuenta["saldo"] + monto
        cuenta["movimientos"].append(("Depósito", monto, cuenta["saldo"]))
        print("Depósito realizado")
    else:
        print("Monto no válido")

def retirar(cuenta, monto):
    if monto <= 0:
        print("Monto no válido")
    elif monto > cuenta["saldo"]:
        print("Saldo insuficiente")
    else:
        cuenta["saldo"] = cuenta["saldo"] - monto
        cuenta["movimientos"].append(("Retiro", monto, cuenta["saldo"]))
        print("Retiro realizado")

while True:
    print("\n1. Consultar saldo")
    print("2. Depositar")
    print("3. Retirar")
    print("4. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        consultar(cuenta)

    elif opcion == "2":
        monto = float(input("Ingrese el monto a depositar: "))
        depositar(cuenta, monto)

    elif opcion == "3":
        monto = float(input("Ingrese el monto a retirar: "))
        retirar(cuenta, monto)

    elif opcion == "4":
        print("\nExtracto de movimientos")
        print("Tipo          Monto          Saldo")

        for movimiento in cuenta["movimientos"]:
            print(movimiento[0], "       ", movimiento[1], "       ", movimiento[2])

        print("Saldo final:", cuenta["saldo"])
        break

    else:
        print("Opción no válida")
```
